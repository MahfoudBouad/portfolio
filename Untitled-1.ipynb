```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the data (adjust path as needed)
path = r"C:\Users\mahfo\OneDrive\Documents\analytic_data2023_0.csv"
df = pd.read_csv(path)

# Select relevant columns
selected_columns = [
    'state', 'county', 
    'v133_rawvalue', 'v083_rawvalue', 'v139_rawvalue',  # Food access
    'v024_rawvalue', 'v063_rawvalue', 'v023_rawvalue',  # Socioeconomic
    'v001_rawvalue', 'v011_rawvalue', 'v060_rawvalue',  # Health outcomes
    'v058_rawvalue',  # % Rural
    'v051_rawvalue'   # Population
]

# Create a clean dataframe with renamed columns
food_df = df[selected_columns].copy()
food_df.columns = [
    'state', 'county',
    'food_env_index', 'limited_food_access', 'food_insecurity',
    'child_poverty', 'median_income', 'unemployment',
    'premature_death', 'adult_obesity', 'diabetes',
    'pct_rural', 'population'
]

# Convert population to numeric (remove commas if present)
food_df['population'] = food_df['population'].replace({',': ''}, regex=True).astype(float)

# Basic data cleaning - drop rows with missing key variables
food_df = food_df.dropna(subset=['limited_food_access', 'food_insecurity', 'premature_death'])

# Display summary statistics
print(food_df.describe())

# Correlation matrix
corr_matrix = food_df[['limited_food_access', 'food_insecurity', 'food_env_index',
                      'child_poverty', 'median_income', 'unemployment',
                      'premature_death', 'adult_obesity', 'diabetes', 'pct_rural']].corr()

plt.figure(figsize=(12, 8))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', center=0)
plt.title('Correlation Matrix of Food Access and Related Factors')
plt.show()
```

```python
# Select food access (LFA_Rate), its probable causes, and consequences columns
FD.columns
numeric_cols = FD[['Poverty_Rate',
       'Median_Income', 'LowAccess_Ratio', 'LowIncome&_LowAccess_Ratio',
       'Pct_LowIncome_', 'NoVehicle_HHs_Ratio', 'SNAP_Participatio_Rate',
       'premature_death', 'poor_health', 'low_birthweight', 'obesity',
       'diabetes', 'unemployment', 'child_poverty']]

# Compute correlation matrix
corr = numeric_cols.corr()

# Plot heatmap
plt.figure(figsize=(12, 8))
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm", linewidths=0.5)
plt.title("Correlation Heatmap of Food Access Variables")
plt.show()
```

```python
# Average health metrics
health_comp = FD.groupby('LowFoodAccess')[['obesity', 'diabetes']].mean()
print("Health Metrics by Food Access:")
print(health_comp)

# Boxplot for obesity
sns.boxplot(data=FD, x='LowFoodAccess', y='obesity', hue="LowFoodAccess")
plt.title('Obesity by Low Food Access')
plt.show()
```
